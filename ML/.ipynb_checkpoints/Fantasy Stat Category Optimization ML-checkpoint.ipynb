```python
import pandas as pd
import numpy as np
from Models.League import *
from constants import *
```

```python
league = fantasyLeague()
```

```python
ratedCats = [cat+"_rating" for cat in mainCats]

data = league.compStatDF[['Year','Week','Team']+ratedCats]
```

```python
data = data.merge(data, on=["Year", "Week"], suffixes=("", "_opp"))
data = data[data["Team"] != data["Team_opp"]]
```

```python
posCats = [cat+"_rating" for cat in mainCats if cat != 'TO']
negCats = ['TO_rating']
posCats_opp = [catRating+"_opp" for catRating in posCats]
negCats_opp = [catRating+"_opp" for catRating in negCats]

data['cats_won'] = (data[posCats].values > data[posCats_opp].values).sum(axis=1) + \
                      (data[negCats].values < data[negCats_opp].values).sum(axis=1)
data['cats_lost'] = (data[posCats].values < data[posCats_opp].values).sum(axis=1) + \
                      (data[negCats].values > data[negCats_opp].values).sum(axis=1)

data['cat_score'] = (data[posCats].values > data[posCats_opp].values).sum(axis=1) + \
                    (data[negCats].values < data[negCats_opp].values).sum(axis=1) + \
                    0.5*(data[posCats].values == data[posCats_opp].values).sum(axis=1)

data['matchup_w'] = data.apply(
    lambda row: 1 if row['cats_won'] > row['cats_lost'] else 
                0 if row['cats_won'] < row['cats_lost'] else 0.75, 
    axis=1
)
data
```

```python
data[data['Year']>=2023].groupby('Team')['cat_score'].mean().sort_values()
```

```python
data.columns
```

```python
focus = posCats+negCats
targets = ['cat_score', 'matchup_w']

dataML = data[focus+targets]
dataML.columns
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# creating this target variable to make it easier to type target column
target = targets[1]

corrMat = dataML.corr()
corrMat.loc[target][focus].plot(kind='barh')
plt.show()
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import RidgeClassifier
from sklearn.feature_selection import SequentialFeatureSelector
```

```python
testSize = 0.3
rand = 400 #arbitrary
multiplier = 100 # need to multiply the target data by this amount to eliminate decimals
X, y = dataML[posCats], 100*dataML[target] #target column multiplied by 2 to get rid of 0.5 decimals
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=testSize, random_state=rand)
```

```python
rr = RidgeClassifier(alpha=10)

sfs = SequentialFeatureSelector(rr, 
                                n_features_to_select=3, 
                                direction="forward",
                                n_jobs=1
                               )

sfs.fit(X_train, y_train)
sfs.get_support()
selected_cols = data.columns[data.columns.isin(posCats)]
predictors = list(selected_cols[sfs.get_support()])
predictors
```

```python
def backtest(data, model, predictors, start=2, step=1):
    all_predictions = []
    
    seasons = sorted(data["Year"].unique())
    
    for i in range(start, len(seasons), step):
        season = seasons[i]
        train = data[data["Year"] < season]
        test = data[data["Year"] == season]
        
        model.fit(train[predictors], multiplier*train[target])
        
        preds = model.predict(test[predictors])
        preds = pd.Series(preds, index=test.index)
        combined = pd.concat([multiplier*test[target], preds], axis=1)
        combined.columns = ["actual", "prediction"]
        
        all_predictions.append(combined)
    return pd.concat(all_predictions)
```

```python
from sklearn.metrics import accuracy_score

predictions = backtest(data, rr, predictors)
acc = accuracy_score(predictions["actual"], predictions["prediction"])
acc
```

```python
from sklearn.pipeline import Pipeline

# Define Ridge Classifier model
rchp = RidgeClassifier()

# Define the Sequential Feature Selector
sfshp = SequentialFeatureSelector(rchp, n_features_to_select='auto', direction='forward')

# Create a pipeline: StandardScaler -> SFS -> RidgeClassifier
pipeline = Pipeline([
    ('sfshp', sfshp),                  # Feature selection
    ('rchp', rchp)        # Ridge Classifier
])

# Define hyperparameter grid
param_grid = {
    'sfshp__n_features_to_select': range(3,5),  # Number of selected features
    'rchp__alpha': [10**x for x in range(-1,7)] # Regularization strength
}

# Set up GridSearchCV
grid = GridSearchCV(pipeline, param_grid, scoring='accuracy', cv=5, n_jobs=-1)

# Fit the model
grid.fit(X_train, y_train)
best_pipeline = grid.best_estimator_

# Best parameters and score
print("Best Parameters:", grid.best_params_)
print("Best Accuracy:", grid.best_score_)
```

```python
best_pipeline.score(X_test, y_test)
```

```python
best_sfs = best_pipeline.named_steps['sfshp']
mask = best_sfs.get_support()

predictors = list(selected_cols[mask])
predictors
```

```python
best_rc = best_pipeline.named_steps['rchp']
coeffs = best_rc.coef_[0]
coeffDict = dict(zip(predictors, coeffs))
coeffDict
```
